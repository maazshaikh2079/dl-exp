```python
# ============================================================
# Experiment 7
# Implement Backpropagation Algorithm to Train a DNN with at least Two Hidden Layers
# ============================================================

import numpy as np

X = np.array([
    [1, 9],
    [2, 5],
    [3, 6]
], dtype=float)

y = np.array([
    [90],
    [85],
    [89]
], dtype=float)

X = X / np.amax(X, axis=0)
y = y / 100


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def derivatives_sigmoid(x):
    return x * (1 - x)


np.random.seed(42)

epochs = 10000
lr = 0.5

inputlayer_neurons = 2
hiddenlayer1_neurons = 3
hiddenlayer2_neurons = 3
output_neurons = 1

W1 = np.random.uniform(
    size=(inputlayer_neurons, hiddenlayer1_neurons)
)
b1 = np.zeros((1, hiddenlayer1_neurons))

W2 = np.random.uniform(
    size=(hiddenlayer1_neurons, hiddenlayer2_neurons)
)
b2 = np.zeros((1, hiddenlayer2_neurons))

W3 = np.random.uniform(
    size=(hiddenlayer2_neurons, output_neurons)
)
b3 = np.zeros((1, output_neurons))

for i in range(epochs):

    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)

    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    z3 = np.dot(a2, W3) + b3
    output = sigmoid(z3)

    error_output = y - output
    d_output = error_output * derivatives_sigmoid(output)

    error_hidden2 = np.dot(d_output, W3.T)
    d_hidden2 = error_hidden2 * derivatives_sigmoid(a2)

    error_hidden1 = np.dot(d_hidden2, W2.T)
    d_hidden1 = error_hidden1 * derivatives_sigmoid(a1)

    W3 += np.dot(a2.T, d_output) * lr
    b3 += np.sum(d_output, axis=0, keepdims=True) * lr

    W2 += np.dot(a1.T, d_hidden2) * lr
    b2 += np.sum(d_hidden2, axis=0, keepdims=True) * lr

    W1 += np.dot(X.T, d_hidden1) * lr
    b1 += np.sum(d_hidden1, axis=0, keepdims=True) * lr

    if (i + 1) % 1000 == 0:
        loss = np.mean((y - output) ** 2)
        print("Epoch:", i + 1, "Loss:", round(loss, 6))

print("\n======================================")
print("FINAL RESULTS")
print("======================================")

print("\nInput:")
print(X)

print("\nActual Output:")
print(y)

print("\nPredicted Output:")
print(output)

predicted_original = output * 100
actual_original = y * 100

print("\nActual Output (Original Scale):")
print(actual_original)

print("\nPredicted Output (Original Scale):")
print(predicted_original)

mse = np.mean((actual_original - predicted_original) ** 2)
rmse = np.sqrt(mse)

print("\nMean Squared Error:", mse)
print("Root Mean Squared Error:", rmse)


# Output:
# Epoch: 1000 Loss: ...
# Epoch: 2000 Loss: ...
# Epoch: 3000 Loss: ...
# Epoch: 4000 Loss: ...
# Epoch: 5000 Loss: ...
# Epoch: 6000 Loss: ...
# Epoch: 7000 Loss: ...
# Epoch: 8000 Loss: ...
# Epoch: 9000 Loss: ...
# Epoch: 10000 Loss: ...
#
# ======================================
# FINAL RESULTS
# ======================================
#
# Input:
# [[0.33333333 1.        ]
#  [0.66666667 0.55555556]
#  [1.         0.66666667]]
#
# Actual Output:
# [[0.9 ]
#  [0.85]
#  [0.89]]
#
# Predicted Output:
# Values depend on the training execution.
#
# Actual Output (Original Scale):
# [[90.]
#  [85.]
#  [89.]]
#
# Predicted Output (Original Scale):
# Values depend on the training execution.
#
# Mean Squared Error: Depends on the predictions
# Root Mean Squared Error: Depends on the predictionsimport numpy as np

X = np.array([
    [1, 9],
    [2, 5],
    [3, 6]
], dtype=float)

y = np.array([
    [90],
    [85],
    [89]
], dtype=float)

X = X / np.amax(X, axis=0)
y = y / 100


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def derivatives_sigmoid(x):
    return x * (1 - x)


np.random.seed(42)

epochs = 10000
lr = 0.5

inputlayer_neurons = 2
hiddenlayer1_neurons = 3
hiddenlayer2_neurons = 3
output_neurons = 1

W1 = np.random.uniform(
    size=(inputlayer_neurons, hiddenlayer1_neurons)
)
b1 = np.zeros((1, hiddenlayer1_neurons))

W2 = np.random.uniform(
    size=(hiddenlayer1_neurons, hiddenlayer2_neurons)
)
b2 = np.zeros((1, hiddenlayer2_neurons))

W3 = np.random.uniform(
    size=(hiddenlayer2_neurons, output_neurons)
)
b3 = np.zeros((1, output_neurons))

for i in range(epochs):

    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)

    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    z3 = np.dot(a2, W3) + b3
    output = sigmoid(z3)

    error_output = y - output
    d_output = error_output * derivatives_sigmoid(output)

    error_hidden2 = np.dot(d_output, W3.T)
    d_hidden2 = error_hidden2 * derivatives_sigmoid(a2)

    error_hidden1 = np.dot(d_hidden2, W2.T)
    d_hidden1 = error_hidden1 * derivatives_sigmoid(a1)

    W3 += np.dot(a2.T, d_output) * lr
    b3 += np.sum(d_output, axis=0, keepdims=True) * lr

    W2 += np.dot(a1.T, d_hidden2) * lr
    b2 += np.sum(d_hidden2, axis=0, keepdims=True) * lr

    W1 += np.dot(X.T, d_hidden1) * lr
    b1 += np.sum(d_hidden1, axis=0, keepdims=True) * lr

    if (i + 1) % 1000 == 0:
        loss = np.mean((y - output) ** 2)
        print("Epoch:", i + 1, "Loss:", round(loss, 6))

print("\n======================================")
print("FINAL RESULTS")
print("======================================")

print("\nInput:")
print(X)

print("\nActual Output:")
print(y)

print("\nPredicted Output:")
print(output)

predicted_original = output * 100
actual_original = y * 100

print("\nActual Output (Original Scale):")
print(actual_original)

print("\nPredicted Output (Original Scale):")
print(predicted_original)

mse = np.mean((actual_original - predicted_original) ** 2)
rmse = np.sqrt(mse)

print("\nMean Squared Error:", mse)
print("Root Mean Squared Error:", rmse)


# Output:
# Epoch: 1000 Loss: ...
# Epoch: 2000 Loss: ...
# Epoch: 3000 Loss: ...
# Epoch: 4000 Loss: ...
# Epoch: 5000 Loss: ...
# Epoch: 6000 Loss: ...
# Epoch: 7000 Loss: ...
# Epoch: 8000 Loss: ...
# Epoch: 9000 Loss: ...
# Epoch: 10000 Loss: ...
#
# ======================================
# FINAL RESULTS
# ======================================
#
# Input:
# [[0.33333333 1.        ]
#  [0.66666667 0.55555556]
#  [1.         0.66666667]]
#
# Actual Output:
# [[0.9 ]
#  [0.85]
#  [0.89]]
#
# Predicted Output:
# Values depend on the training execution.
#
# Actual Output (Original Scale):
# [[90.]
#  [85.]
#  [89.]]
#
# Predicted Output (Original Scale):
# Values depend on the training execution.
#
# Mean Squared Error: Depends on the predictions
# Root Mean Squared Error: Depends on the predictions
```
