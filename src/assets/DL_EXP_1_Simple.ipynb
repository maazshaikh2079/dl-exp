```python
# Experiment No. 1 (Simple)
# Aim: To explore Python libraries for Deep Learning (Keras, TensorFlow)

# Description:
# This program uses TensorFlow and Keras to create a simple neural network.
# It trains the model using sample input data to learn the AND gate operation.
# It demonstrates how deep learning models are created and trained in Python.

import tensorflow as tf
from tensorflow import keras
import numpy as np

# Training data
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=np.float32)
y = np.array([[0], [0], [0], [1]], dtype=np.float32)

# Create neural network model
model = keras.Sequential([
    keras.Input(shape=(2,)),
    keras.layers.Dense(8, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid')
])

# Compile model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train model
for epoch in range(100):
    with tf.GradientTape() as tape:
        predictions = model(X, training=True)
        loss = keras.losses.binary_crossentropy(y, predictions)
        loss = tf.reduce_mean(loss)

    gradients = tape.gradient(loss, model.trainable_variables)
    model.optimizer.apply_gradients(
        zip(gradients, model.trainable_variables)
    )

# Display results
predictions = model(X, training=False)

predicted_values = (
    tf.keras.backend.get_value(predictions) >= 0.5
).astype(int)

accuracy = np.mean(predicted_values == y) * 100

print("TensorFlow Version:", tf.__version__)

print("\nInput Data:")
print(X.astype(int))

print("\nActual Output:")
print(y.astype(int).flatten())

print("\nPredicted Output:")
print(predicted_values.flatten())

print("\nModel Accuracy:", accuracy, "%")

# Output:
# TensorFlow Version: (Installed version)
#
# Input Data:
# [[0 0]
#  [0 1]
#  [1 0]
#  [1 1]]
#
# Actual Output:
# [0 0 0 1]
#
# Predicted Output:
# [0 0 0 1]
#
# Model Accuracy: 100.0 %
```
