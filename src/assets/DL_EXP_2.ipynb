```python
# AND Gate
def mcp_neuron_AND(x1, x2, w1=1, w2=1, threshold=2):
    summation = x1 * w1 + x2 * w2

    if summation >= threshold:
        return 1
    else:
        return 0


inputs = [
    (0, 0),
    (0, 1),
    (1, 0),
    (1, 1)
]
print("AND Gate")
print("x1 x2 Output")

for x1, x2 in inputs:
    y = mcp_neuron_AND(x1, x2)
    print(x1, x2, " ", y)


```

```python
# NOT GATE
def NOT(x):
    weight = -1
    threshold = 0

    summation = x * weight

    if summation >= threshold:
        return 1
    else:
        return 0


print("\nNOT Gate")
print("x Output")

for i in [0, 1]:
    print(i, " ", NOT(i))
```

```python
# OR GATE
def OR(x1, x2, w1=1, w2=1, threshold=1):
    summation = x1 * w1 + x2 * w2

    if summation >= threshold:
        return 1
    else:
        return 0


inputs = [
    (0, 0),
    (0, 1),
    (1, 0),
    (1, 1)
]
print("OR Gate")
print("x1 x2 Output")

for x1, x2 in inputs:
    y = OR(x1, x2)
    print(x1, x2, " ", y)

```

```python
# NOR gate
def NOR(x1, x2, w1=-1, w2=-1, threshold=0):
    summation = x1*w1 + x2*w2

    if summation >= threshold:
        return 1
    else:
        return 0


inputs = [(0,0),(0,1),(1,0),(1,1)]

print("NOR Gate")
print("x1 x2 Output")

for x1, x2 in inputs:
    print(x1, x2, " ", NOR(x1,x2))
```

```python
# Expt. No. 2 (Manual)
# Title: To implement McCulloch Pitts model for binary logic functions

import numpy as np

class McCullochPittsNeuron:
    def __init__(self, weights, threshold):
        self.weights = weights
        self.threshold = threshold

    def activate(self, inputs):
        weighted_sum = np.dot(inputs, self.weights)
        return 1 if weighted_sum >= self.threshold else 0


def AND(inputs):
    weights = [1, 1]
    threshold = 2
    neuron = McCullochPittsNeuron(weights, threshold)
    return neuron.activate(inputs)


def OR(inputs):
    weights = [1, 1]
    threshold = 1
    neuron = McCullochPittsNeuron(weights, threshold)
    return neuron.activate(inputs)


def NOT(inputs):
    weights = [-1]
    threshold = -0.5
    neuron = McCullochPittsNeuron(weights, threshold)
    return neuron.activate(inputs)


print("AND Gate")
print("0 AND 0:", AND([0, 0]))
print("0 AND 1:", AND([0, 1]))
print("1 AND 0:", AND([1, 0]))
print("1 AND 1:", AND([1, 1]))

print("\nOR Gate")
print("0 OR 0:", OR([0, 0]))
print("0 OR 1:", OR([0, 1]))
print("1 OR 0:", OR([1, 0]))
print("1 OR 1:", OR([1, 1]))

print("\nNOT Gate")
print("NOT 0:", NOT([0]))
print("NOT 1:", NOT([1]))


# OUTPUT:
# AND Gate
# 0 AND 0: 0
# 0 AND 1: 0
# 1 AND 0: 0
# 1 AND 1: 1
#
# OR Gate
# 0 OR 0: 0
# 0 OR 1: 1
# 1 OR 0: 1
# 1 OR 1: 1
#
# NOT Gate
# NOT 0: 1
# NOT 1: 0
```
